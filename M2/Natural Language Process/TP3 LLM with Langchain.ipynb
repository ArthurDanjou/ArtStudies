
## LangChain Expression Language (LCEL) 🔗

### What is LCEL? 🤔

LCEL is LangChain's way of chaining components together using the `|` (pipe) operator.

```python
# Traditional approach
formatted = prompt.format(...)
response = model.invoke(formatted)

# LCEL approach ✨
chain = prompt | model
response = chain.invoke({...})
```

### Why Use LCEL? 🌟

| Feature | Benefit |
|---------|--------|
| **Composability** | Chain multiple operations easily |
| **Streaming** | Automatic streaming support |
| **Async** | Built-in async execution |
| **Batching** | Process multiple inputs efficiently |
| **Debugging** | Better error messages and tracing |

### Visual Representation

```
Input ──> Prompt ──> Model ──> Output Parser ──> Final Result
          Template
          
With LCEL:
chain = prompt | model | output_parser
result = chain.invoke(input)
```

## 📚 Exercise Set 1: Chat Model Fundamentals


````python
from __future__ import annotations

from typing import Any

from langchain_mistralai import ChatMistralAI

MODEL_PRICING_PER_1K_TOKENS = {
    "mistral-small-latest": {"input": 0.0002, "output": 0.0006},
    "mistral-large-latest": {"input": 0.002, "output": 0.006},
    "ministral-3b-2512": {"input": 0.00004, "output": 0.00012},
}


def _extract_token_usage(response: Any) -> dict[str, int]:
    usage = response.response_metadata.get("token_usage", {})
    prompt_tokens = int(usage.get("prompt_tokens", 0))
    completion_tokens = int(
        usage.get("completion_tokens", usage.get("output_tokens", 0)),
    )
    total_tokens = int(usage.get("total_tokens", prompt_tokens + completion_tokens))
    return {
        "prompt_tokens": prompt_tokens,
        "completion_tokens": completion_tokens,
        "total_tokens": total_tokens,
    }


def _estimate_cost(
    model_name: str,
    prompt_tokens: int,
    completion_tokens: int,
) -> float:
    pricing = MODEL_PRICING_PER_1K_TOKENS.get(
        model_name,
        MODEL_PRICING_PER_1K_TOKENS["mistral-small-latest"],
    )
    return (prompt_tokens / 1000) * pricing["input"] + (
        completion_tokens / 1000
    ) * pricing["output"]


def compare_model_responses(
    prompt: str,
    models: list[str],
) -> dict[str, dict[str, Any]]:
    """Send the same prompt to multiple models and compare responses."""
    results: dict[str, dict[str, Any]] = {}

    for model_name in models:
        llm = ChatMistralAI(model=model_name, temperature=0)
        response = llm.invoke(prompt)
        usage = _extract_token_usage(response)
        cost = _estimate_cost(
            model_name,
            usage["prompt_tokens"],
            usage["completion_tokens"],
        )

        results[model_name] = {
            "content": response.content,
            "token_count": usage["total_tokens"],
            "prompt_tokens": usage["prompt_tokens"],
            "completion_tokens": usage["completion_tokens"],
            "cost": cost,
            "model_name": model_name,
        }

    return results


test_prompt = "Explain quantum computing in one paragraph"
model_names = ["mistral-small-latest", "mistral-large-latest"]

results = compare_model_responses(test_prompt, model_names)

for model_name, response_data in results.items():
    print(f"\n{'=' * 70}")
    print(f"Model: {model_name}")
    print(f"{'=' * 70}")
    print(f"Response: {response_data['content']}")
    print(f"Tokens used: {response_data['token_count']}")
    print(f"Cost: ${response_data['cost']:.6f}")
````

````python
from langchain_mistralai import ChatMistralAI


def calculate_variety(responses: list[str]) -> float:
    """Simple lexical diversity score in [0, 1]."""
    words: list[str] = []
    for text in responses:
        words.extend(text.lower().split())
    if not words:
        return 0.0
    unique_words = len(set(words))
    return round(unique_words / len(words), 3)


def _content_to_text(content: object) -> str:
    if isinstance(content, str):
        return content
    if isinstance(content, list):
        parts: list[str] = []
        for item in content:
            if isinstance(item, str):
                parts.append(item)
            elif isinstance(item, dict):
                text = item.get("text", "")
                parts.append(text if isinstance(text, str) else str(text))
            else:
                parts.append(str(item))
        return " ".join(part for part in parts if part).strip()
    return str(content)


def temperature_experiment(
    prompt: str,
    temperatures: list[float] = [0.0, 0.5, 1.0],
) -> dict[float, list[str]]:
    """Test how temperature affects model creativity."""
    results: dict[float, list[str]] = {}

    for temp in temperatures:
        llm = ChatMistralAI(model="mistral-small-latest", temperature=temp)
        temp_outputs: list[str] = []
        for _ in range(3):
            response = llm.invoke(prompt)
            temp_outputs.append(_content_to_text(response.content))
        results[temp] = temp_outputs

    return results


creative_prompt = "Write a two-sentence story about a robot learning to paint"

results = temperature_experiment(creative_prompt)

print("🌡️ Temperature Impact Analysis\n")
for temp, responses in results.items():
    print(f"\nTemperature: {temp}")
    print(f"Response 1: {responses[0]}")
    print(f"Response 2: {responses[1]}")
    print(f"Response 3: {responses[2]}")
    print(f"Variety Score: {calculate_variety(responses)}")
````

````python
import time

from langchain_mistralai import ChatMistralAI


def analyze_response_metadata(
    prompt: str,
    model_name: str = "mistral-small-latest",
) -> dict[str, Any]:
    """Analyze metadata from a model response."""
    llm = ChatMistralAI(model=model_name, temperature=0)

    start = time.perf_counter()
    response = llm.invoke(prompt)
    elapsed = time.perf_counter() - start

    usage = response.response_metadata.get("token_usage", {})
    prompt_tokens = int(usage.get("prompt_tokens", 0))
    completion_tokens = int(
        usage.get("completion_tokens", usage.get("output_tokens", 0)),
    )
    total_tokens = int(usage.get("total_tokens", prompt_tokens + completion_tokens))

    pricing = {
        "mistral-small-latest": {"input": 0.0002, "output": 0.0006},
        "mistral-large-latest": {"input": 0.002, "output": 0.006},
        "ministral-3b-2512": {"input": 0.00004, "output": 0.00012},
    }.get(model_name, {"input": 0.0002, "output": 0.0006})

    prompt_cost = (prompt_tokens / 1000) * pricing["input"]
    completion_cost = (completion_tokens / 1000) * pricing["output"]
    total_cost = prompt_cost + completion_cost

    return {
        "model_name": response.response_metadata.get("model", model_name),
        "prompt_tokens": prompt_tokens,
        "completion_tokens": completion_tokens,
        "total_tokens": total_tokens,
        "finish_reason": response.response_metadata.get("stop_reason", "unknown"),
        "prompt_cost": prompt_cost,
        "completion_cost": completion_cost,
        "cost": total_cost,
        "response_time": elapsed,
    }


test_prompts = [
    "Hi",
    "Explain machine learning in 100 words",
    "Write a detailed analysis of the impact of AI on healthcare" * 10,
]

for prompt in test_prompts:
    metadata = analyze_response_metadata(prompt)
    print(f"\nPrompt length: {len(prompt)} chars")
    print(
        f"Tokens: {metadata['total_tokens']} (in: {metadata['prompt_tokens']}, out: {metadata['completion_tokens']})",
    )
    print(f"Cost: ${metadata['cost']:.6f}")
    print(f"Time: {metadata['response_time']:.2f}s")
````

## 💬 Exercise Set 2: Message Roles and Conversations


````python
from langchain_mistralai import ChatMistralAI


def optimize_system_prompt(
    user_query: str,
    system_prompts: list[str],
    evaluation_criteria: str = "helpful, clear, empathetic",
) -> dict[str, Any]:
    """Test multiple system prompts and rank them with a simple heuristic score."""
    judge = ChatMistralAI(model="mistral-small-latest", temperature=0)
    results: list[dict[str, Any]] = []

    for sys_prompt in system_prompts:
        candidate = judge.invoke(
            [
                {"role": "system", "content": sys_prompt},
                {"role": "user", "content": user_query},
            ],
        ).content

        evaluation_prompt = f"""
        Evaluate this support response with criteria: {evaluation_criteria}.
        Return only an integer score from 0 to 100.

        USER QUERY: {user_query}
        RESPONSE: {candidate}
        """
        raw_score = judge.invoke(evaluation_prompt).content.strip()
        digits = "".join(ch for ch in raw_score if ch.isdigit())
        score = int(digits) if digits else 60
        score = max(0, min(score, 100))

        results.append(
            {
                "system_prompt": sys_prompt,
                "response": candidate,
                "score": score,
            },
        )

    ranked = sorted(results, key=lambda item: item["score"], reverse=True)
    return {"ranking": ranked, "best": ranked[0] if ranked else None}


user_question = "My order hasn't arrived yet and it's been 2 weeks"

system_prompt_variations = [
    "You are a customer support agent.",
    "You are an empathetic customer support agent who prioritizes customer satisfaction.",
    "You are a customer service representative. Be concise, helpful, and solution-oriented.",
    "You are a friendly support agent. Use a warm tone and always offer multiple solutions.",
]

results = optimize_system_prompt(
    user_question,
    system_prompt_variations,
    evaluation_criteria="empathy, clarity, actionability",
)

print("Top system prompt:")
print(results["best"]["system_prompt"])
print("Score:", results["best"]["score"])
````

````python
from langchain_core.messages import AIMessage, HumanMessage, SystemMessage
from langchain_mistralai import ChatMistralAI


class ConversationManager:
    """Manage multi-turn conversations with memory."""

    def __init__(self, system_prompt: str, model_name: str = "mistral-small-latest"):
        self.system_prompt = system_prompt
        self.model = ChatMistralAI(model=model_name, temperature=0.7)
        self.history: list[Any] = []
        self.total_tokens = 0

    def send_message(self, user_message: str) -> str:
        self.history.append(HumanMessage(content=user_message))
        messages = [SystemMessage(content=self.system_prompt), *self.history]

        response = self.model.invoke(messages)
        self.history.append(AIMessage(content=response.content))

        usage = response.response_metadata.get("token_usage", {})
        self.total_tokens += int(
            usage.get(
                "total_tokens",
                usage.get("prompt_tokens", 0) + usage.get("completion_tokens", 0),
            ),
        )
        return response.content

    def get_conversation_summary(self) -> dict[str, Any]:
        human_count = sum(
            1 for msg in self.history if getattr(msg, "type", "") == "human"
        )
        ai_count = sum(1 for msg in self.history if getattr(msg, "type", "") == "ai")
        return {
            "message_count": len(self.history),
            "user_messages": human_count,
            "assistant_messages": ai_count,
            "total_tokens": self.total_tokens,
        }

    def clear_history(self):
        self.history = []
        self.total_tokens = 0


manager = ConversationManager(
    system_prompt="You are a helpful Python tutor. Explain concepts clearly with examples.",
)

conversation_flow = [
    "What is a list in Python?",
    "How do I add items to a list?",
    "What's the difference between append and extend?",
    "Can you show me an example?",
]

for user_msg in conversation_flow:
    response = manager.send_message(user_msg)
    print(f"\n👤 User: {user_msg}")
    print(f"🤖 Assistant: {response}")

summary = manager.get_conversation_summary()
print("\n📊 Conversation Summary:")
print(f"Messages: {summary['message_count']}")
print(f"Total tokens: {summary['total_tokens']}")
````

````python
from langchain_core.messages import HumanMessage, SystemMessage


def create_conversation_with_roles(scenario: str) -> list:
    """Create a conversation with multiple roles."""
    if scenario == "interview":
        return [
            SystemMessage(
                content="You are a technical interviewer for a Python backend role. Ask focused questions.",
            ),
            HumanMessage(content="Candidate is ready for the interview."),
            HumanMessage(content="Can you ask the first interview question?"),
            HumanMessage(content="Ask a follow-up question about API design."),
        ]

    if scenario == "debate":
        return [
            SystemMessage(
                content="You moderate a short debate and provide balanced viewpoints.",
            ),
            HumanMessage(
                content="Debate topic: Is remote work better than office work?",
            ),
            HumanMessage(
                content="Provide argument A and argument B, then a neutral synthesis.",
            ),
        ]

    if scenario == "tutoring":
        return [
            SystemMessage(
                content="You are a patient tutor. Increase difficulty gradually.",
            ),
            HumanMessage(content="What is a variable in Python?"),
            HumanMessage(content="Now explain functions with a simple example."),
            HumanMessage(content="Finally, explain decorators at an intuitive level."),
        ]

    return [
        SystemMessage(content="You are a helpful assistant."),
        HumanMessage(content="Start a useful conversation."),
    ]


conversation = create_conversation_with_roles("interview")

model = ChatMistralAI(model="mistral-small-latest")
for idx, msg in enumerate(conversation):
    if msg.type == "human":
        context = conversation[: idx + 1]
        response = model.invoke(context)
        print(f"\n💬 human: {msg.content}")
        print(f"🤖 AI: {response.content}")
````

## 🌊 Exercise Set 3: Streaming and Real-Time Responses


````python
from langchain_mistralai import ChatMistralAI


class StreamingHandler:
    """Handle streaming responses with additional features."""

    def __init__(self):
        self.full_response = ""
        self.chunk_count = 0
        self.start_time = None

    def stream_with_stats(self, model, prompt: str) -> dict:
        self.full_response = ""
        self.chunk_count = 0
        self.start_time = time.perf_counter()

        for chunk in model.stream(prompt):
            content = chunk.content or ""
            self.full_response += content
            self.chunk_count += 1
            print(content, end="")

        total_time = time.perf_counter() - self.start_time
        avg_chunk_time = total_time / max(self.chunk_count, 1)
        return {
            "response": self.full_response,
            "chunk_count": self.chunk_count,
            "total_time": total_time,
            "avg_chunk_time": avg_chunk_time,
        }

    def stream_with_typewriter_effect(self, model, prompt: str, delay: float = 0.03):
        self.full_response = ""
        for chunk in model.stream(prompt):
            content = chunk.content or ""
            self.full_response += content
            print(content, end="", flush=True)
            time.sleep(delay)
        return self.full_response


handler = StreamingHandler()
model = ChatMistralAI(model="mistral-small-latest", streaming=True)

prompt = "Write a short story about a time traveler"
stats = handler.stream_with_stats(model, prompt)

print("\n📊 Streaming Statistics:")
print(f"Chunks received: {stats['chunk_count']}")
print(f"Total time: {stats['total_time']:.2f}s")
print(f"Avg chunk time: {stats['avg_chunk_time']:.3f}s")
````

````python
import asyncio

from langchain_mistralai import ChatMistralAI


async def parallel_chat_requests(
    prompts: list[str],
    model_name: str = "mistral-small-latest",
) -> list[dict]:
    """Send multiple prompts in parallel for efficiency."""
    llm = ChatMistralAI(model=model_name, temperature=0)

    async def _one(prompt: str) -> dict:
        start = time.perf_counter()
        response = await llm.ainvoke(prompt)
        elapsed = time.perf_counter() - start
        usage = response.response_metadata.get("token_usage", {})
        total_tokens = int(
            usage.get(
                "total_tokens",
                usage.get("prompt_tokens", 0) + usage.get("completion_tokens", 0),
            ),
        )
        return {
            "prompt": prompt,
            "response": response.content,
            "time": elapsed,
            "tokens": total_tokens,
        }

    tasks = [_one(prompt) for prompt in prompts]
    return await asyncio.gather(*tasks)


translation_prompts = [
    "Translate 'Hello, how are you?' to French",
    "Translate 'Hello, how are you?' to Spanish",
    "Translate 'Hello, how are you?' to German",
    "Translate 'Hello, how are you?' to Italian",
    "Translate 'Hello, how are you?' to Japanese",
]

results = await parallel_chat_requests(translation_prompts)

print("⚡ Parallel Execution Results:")
total_time = sum(r["time"] for r in results)
print(f"Total time (parallel): {max(r['time'] for r in results):.2f}s")
print(f"Time if sequential: {total_time:.2f}s")
print(f"Speedup: {total_time / max(r['time'] for r in results):.1f}x")
````

````python
from langchain_core.prompts import ChatPromptTemplate
from langchain_mistralai import ChatMistralAI


def create_content_pipeline():
    model = ChatMistralAI(model="mistral-small-latest", temperature=0.6)
    parser = StrOutputParser()

    topic_chain = (
        ChatPromptTemplate.from_template(
            "Generate one compelling blog title on: {subject} (tone: {tone}).",
        )
        | model
        | parser
    )
    outline_chain = (
        ChatPromptTemplate.from_template(
            "Create a concise outline for the blog title: {title}. Include 4 sections.",
        )
        | model
        | parser
    )
    intro_chain = (
        ChatPromptTemplate.from_template(
            "Write an introduction ({intro_words} words) for title '{title}' with outline:\n{outline}",
        )
        | model
        | parser
    )
    conclusion_chain = (
        ChatPromptTemplate.from_template(
            "Write a conclusion ({conclusion_words} words) for title '{title}' with outline:\n{outline}",
        )
        | model
        | parser
    )

    def run(payload: dict) -> str:
        title = topic_chain.invoke(payload)
        outline = outline_chain.invoke({"title": title})
        total_words = int(payload.get("word_count", 500))
        intro_words = max(80, total_words // 5)
        conclusion_words = max(80, total_words // 6)
        intro = intro_chain.invoke(
            {
                "title": title,
                "outline": outline,
                "intro_words": intro_words,
            },
        )
        conclusion = conclusion_chain.invoke(
            {
                "title": title,
                "outline": outline,
                "conclusion_words": conclusion_words,
            },
        )
        return (
            f"# {title}\n\n"
            f"## Outline\n{outline}\n\n"
            f"## Introduction\n{intro}\n\n"
            f"## Conclusion\n{conclusion}"
        )

    return RunnableLambda(run)


result = create_content_pipeline().invoke(
    {
        "subject": "artificial intelligence in healthcare",
        "tone": "professional but accessible",
        "word_count": 500,
    },
)

print("📝 Generated Content:")
print(result)
````

## 🔀 Exercise Set 6: Prompt Routing & Dynamic Workflows

Routing directs inputs to different processing paths based on content, allowing intelligent workflow management.

````python
import json

from langchain_core.prompts import ChatPromptTemplate
from langchain_mistralai import ChatMistralAI


def create_email_router():
    model = ChatMistralAI(model="mistral-small-latest", temperature=0.1)

    classifier_prompt = ChatPromptTemplate.from_template(
        "Classify this email. Return ONLY a JSON with keys urgency, category, sentiment:\n{email}",
    )
    classifier_chain = classifier_prompt | model | StrOutputParser()

    def _parse_classification(raw: str) -> dict:
        try:
            return json.loads(raw)
        except json.JSONDecodeError:
            text = raw.lower()
            urgency = (
                "urgent" if "urgent" in text or "losing money" in text else "normal"
            )
            if "api" in text or "error" in text:
                category = "technical"
            elif "price" in text or "charged" in text:
                category = "inquiry"
            elif "amazing" in text or "great" in text:
                category = "feedback"
            else:
                category = "inquiry"
            sentiment = "negative" if "!" in text or "error" in text else "neutral"
            return {"urgency": urgency, "category": category, "sentiment": sentiment}

    def route(payload: dict) -> str:
        email = payload["email"]
        classification = _parse_classification(
            classifier_chain.invoke({"email": email}),
        )

        urgency = classification.get("urgency", "normal")
        category = classification.get("category", "inquiry")
        sentiment = classification.get("sentiment", "neutral")

        if urgency == "urgent" and category in {"complaint", "technical"}:
            strategy = "Prioritize immediate apology + concrete ETA + escalation path."
        elif category == "technical":
            strategy = "Provide troubleshooting checklist and request key logs."
        elif category == "feedback":
            strategy = "Thank user warmly and invite optional testimonial."
        else:
            strategy = "Provide concise answer and link to relevant resources."

        return (
            f"Classification: {classification}\n"
            f"Suggested strategy: {strategy}\n"
            f"Tone: {sentiment}"
        )

    return RunnableLambda(route)


test_emails = [
    "URGENT: Your service has been down for 3 hours! We're losing money!",
    "Hi, I'm curious about your pricing plans for enterprises.",
    "Just wanted to say your product is amazing! Keep up the great work!",
    "How do I integrate your API with Python? Getting error 403.",
]

router = create_email_router()
for email in test_emails:
    result = router.invoke({"email": email})
    print(f"\n📧 Email: {email[:50]}...")
    print(f"Response: {result}")
````

## 🎯 Exercise Set 8: Orchestrator-Worker Patterns

### What is the Orchestrator-Worker Pattern? 🤔

This pattern divides a complex task among specialized workers:
- **Orchestrator**: The "manager" that coordinates workflow, decides task order, and combines results
- **Workers**: Specialized "experts" that each handle one specific type of task

### Key Concepts:
1. **Separation of Concerns**: Each worker has ONE job and does it well
2. **Sequential vs Parallel**: Orchestrator decides which tasks can run in parallel vs must be sequential
3. **State Management**: Orchestrator maintains context and shares relevant info with workers
4. **Result Integration**: Orchestrator combines worker outputs into final result

### Architecture Pattern:
```
                    [Orchestrator]
                          |
        +-----------------+-----------------+
        |                 |                 |
    [Worker A]        [Worker B]        [Worker C]
    (Specialist)      (Specialist)      (Specialist)
        |                 |                 |
        +-----------------+-----------------+
                          |
                    [Final Result]
```

Let's build some orchestrator systems! Each exercise has detailed steps below.

### 📝 Exercise 8.2: Content Creation Orchestrator

**Goal**: Build a system that creates high-quality articles through multiple specialized workers.

**Key Learning**: How to combine parallel and sequential stages in orchestration.

**Workflow Strategy**:
- **Phase 1** (Parallel): Multiple researchers gather different types of information
- **Phase 2** (Sequential): Writer creates draft from research
- **Phase 3** (Parallel): Editor, SEO specialist, and fact-checker review simultaneously
- **Phase 4** (Sequential): Integrate all feedback into final article

````python
from langchain_core.prompts import ChatPromptTemplate
from langchain_mistralai import ChatMistralAI


class ContentOrchestrator:
    """Orchestrate content creation with specialized workers."""

    def __init__(self):
        self.model = ChatMistralAI(model="mistral-small-latest", temperature=0.6)
        parser = StrOutputParser()

        self.workers = {
            "researcher": (
                ChatPromptTemplate.from_messages(
                    [
                        ("system", "You are a research specialist."),
                        ("user", "Research topic: {topic} for audience: {audience}"),
                    ],
                )
                | self.model
                | parser
            ),
            "trend_analyst": (
                ChatPromptTemplate.from_messages(
                    [
                        ("system", "You are a trend analyst."),
                        ("user", "Analyze trends for topic: {topic}"),
                    ],
                )
                | self.model
                | parser
            ),
            "writer": (
                ChatPromptTemplate.from_messages(
                    [
                        ("system", "You are a professional writer."),
                        (
                            "user",
                            "Write an article draft on {topic} for {audience}. Use this research:\n{research}",
                        ),
                    ],
                )
                | self.model
                | parser
            ),
            "editor": (
                ChatPromptTemplate.from_messages(
                    [
                        ("system", "You are an editor improving clarity and flow."),
                        ("user", "Edit this draft:\n{content}"),
                    ],
                )
                | self.model
                | parser
            ),
            "seo": (
                ChatPromptTemplate.from_messages(
                    [
                        ("system", "You are an SEO specialist."),
                        (
                            "user",
                            "Optimize this article for SEO on topic '{topic}':\n{content}",
                        ),
                    ],
                )
                | self.model
                | parser
            ),
            "fact_checker": (
                ChatPromptTemplate.from_messages(
                    [
                        ("system", "You are a fact-checker."),
                        (
                            "user",
                            "Review factual claims and flag uncertainty:\n{content}",
                        ),
                    ],
                )
                | self.model
                | parser
            ),
        }

    async def create_article(self, topic: str, target_audience: str):
        results = {}

        research_tasks = [
            self.workers["researcher"].ainvoke(
                {"topic": topic, "audience": target_audience},
            ),
            self.workers["trend_analyst"].ainvoke({"topic": topic}),
        ]
        research_general, research_trends = await asyncio.gather(*research_tasks)
        combined_research = (
            f"General research:\n{research_general}\n\nTrends:\n{research_trends}"
        )
        results["research_sources"] = ["general_research", "trend_analysis"]

        draft = await self.workers["writer"].ainvoke(
            {
                "topic": topic,
                "audience": target_audience,
                "research": combined_research,
            },
        )

        review_tasks = [
            self.workers["editor"].ainvoke({"content": draft}),
            self.workers["seo"].ainvoke({"content": draft, "topic": topic}),
            self.workers["fact_checker"].ainvoke({"content": draft}),
        ]
        edited, seo_feedback, fact_feedback = await asyncio.gather(*review_tasks)

        final_content = f"{edited}\n\n---\nSEO Notes:\n{seo_feedback}\n\n---\nFact-check Notes:\n{fact_feedback}"

        results.update(
            {
                "title": f"{topic}: Practical Insights",
                "content": final_content,
                "word_count": len(final_content.split()),
                "seo_score": min(100, 70 + len(seo_feedback.split()) // 8),
                "fact_check_status": "reviewed",
                "keywords": [topic.split()[0].lower(), "future", "strategy"],
            },
        )
        return results


content_orch = ContentOrchestrator()
result = await content_orch.create_article(
    topic="The Future of Remote Work",
    target_audience="business professionals",
)

print("📰 Article Creation Results:")
print(f"Title: {result['title']}")
print(f"Word Count: {result['word_count']}")
print(f"SEO Score: {result['seo_score']}/100")
print(f"Fact-check Status: {result['fact_check_status']}")
print(f"\nFirst paragraph:\n{result['content'][:200]}...")
````

### 🎫 Exercise 8.3: Customer Support Orchestrator

**Goal**: Build an intelligent support system that analyzes tickets and routes to appropriate handlers.

**Key Learning**: Decision-making in orchestration - when to use which workers based on ticket analysis.

**Decision Tree**:
```
Ticket Received
    → Triage (categorize + assess complexity)
    → IF simple → Use KB Worker → Generate Response
    → IF technical → Use Technical Worker + KB Worker (parallel) → Generate Response
    → IF complex → Use All Workers + Escalation Check → Generate Response
    → Create Follow-up Plan
```

````python
from langchain_core.prompts import ChatPromptTemplate
from langchain_mistralai import ChatMistralAI


class SupportOrchestrator:
    """Orchestrate customer support with specialized workers."""

    def __init__(self):
        self.model = ChatMistralAI(model="mistral-small-latest", temperature=0.3)
        parser = StrOutputParser()

        self.workers = {
            "triage": (
                ChatPromptTemplate.from_messages(
                    [
                        ("system", "You are a support triage specialist."),
                        (
                            "user",
                            "Ticket subject: {subject}\nMessage: {message}\nPriority: {priority}\nReturn: category, urgency, complexity, key_issue",
                        ),
                    ],
                )
                | self.model
                | parser
            ),
            "kb": (
                ChatPromptTemplate.from_messages(
                    [
                        ("system", "You are a support knowledge base assistant."),
                        ("user", "Provide KB-based help for: {message}"),
                    ],
                )
                | self.model
                | parser
            ),
            "technical": (
                ChatPromptTemplate.from_messages(
                    [
                        ("system", "You are a technical support engineer."),
                        ("user", "Diagnose likely root cause and fix for: {message}"),
                    ],
                )
                | self.model
                | parser
            ),
            "escalation": (
                ChatPromptTemplate.from_messages(
                    [
                        ("system", "You decide whether escalation to human is needed."),
                        ("user", "Ticket + analysis:\n{analysis}"),
                    ],
                )
                | self.model
                | parser
            ),
            "response_writer": (
                ChatPromptTemplate.from_messages(
                    [
                        ("system", "Write a professional customer-facing response."),
                        (
                            "user",
                            "Ticket:\n{ticket}\nAnalysis:\n{analysis}\nSolutions:\n{solutions}\nEscalate: {escalate}",
                        ),
                    ],
                )
                | self.model
                | parser
            ),
        }

    async def handle_ticket(self, ticket: dict):
        triage_raw = await self.workers["triage"].ainvoke(ticket)
        triage_text = triage_raw.lower()

        if "technical" in triage_text or "401" in ticket["message"]:
            category = "technical"
        elif "billing" in triage_text or "charged" in ticket["message"].lower():
            category = "billing"
        else:
            category = "general"

        if "complex" in triage_text or "high" in triage_text:
            complexity = "complex"
        elif "moderate" in triage_text:
            complexity = "moderate"
        else:
            complexity = "simple"

        kb_task = self.workers["kb"].ainvoke({"message": ticket["message"]})

        if complexity == "simple":
            kb_result = await kb_task
            combined_solutions = kb_result
            workers_used = ["triage", "kb"]
        else:
            technical_task = self.workers["technical"].ainvoke(
                {"message": ticket["message"]},
            )
            kb_result, technical_result = await asyncio.gather(kb_task, technical_task)
            combined_solutions = f"KB:\n{kb_result}\n\nTechnical:\n{technical_result}"
            workers_used = ["triage", "kb", "technical"]

        escalate = (
            complexity == "complex" or ticket.get("priority", "").lower() == "high"
        )
        escalation_note = "No escalation required."
        if escalate:
            escalation_note = await self.workers["escalation"].ainvoke(
                {"analysis": f"Triage: {triage_raw}\nSolutions: {combined_solutions}"},
            )
            workers_used.append("escalation")

        response = await self.workers["response_writer"].ainvoke(
            {
                "ticket": ticket,
                "analysis": triage_raw,
                "solutions": combined_solutions,
                "escalate": escalate,
            },
        )
        workers_used.append("response_writer")

        return {
            "category": category,
            "root_cause": escalation_note
            if escalate
            else "Likely authentication/configuration mismatch",
            "solution": combined_solutions,
            "response": response,
            "escalate": escalate,
            "eta": "15 minutes" if not escalate else "2-4 hours",
            "confidence": 0.9 if not escalate else 0.72,
            "workers_used": workers_used,
        }


support_orch = SupportOrchestrator()
ticket = {
    "customer_id": "C12345",
    "subject": "Cannot connect to API - getting 401 errors",
    "message": "I've been trying to use your API for the past hour but keep getting 401 Unauthorized. I'm using the API key from my dashboard.",
    "priority": "high",
}

result = await support_orch.handle_ticket(ticket)

print("🎫 Support Ticket Resolution:")
print(f"Category: {result['category']}")
print(f"Root Cause: {result['root_cause']}")
print(f"Solution: {result['solution']}")
print(f"Escalate: {result['escalate']}")
print(f"Estimated Resolution Time: {result['eta']}")
````

## 🔄 Exercise Set 9: Evaluator-Optimizer Loops

Implement iterative improvement systems where an evaluator judges output quality and an optimizer refines it.

````python
from langchain_core.prompts import ChatPromptTemplate
from langchain_mistralai import ChatMistralAI


class EssayOptimizerLoop:
    """Iteratively improve essays through evaluate-optimize cycles."""

    def __init__(self, max_iterations=3, target_score=85):
        self.model = ChatMistralAI(model="mistral-small-latest", temperature=0.7)
        self.max_iterations = max_iterations
        self.target_score = target_score
        self.parser = StrOutputParser()

    async def generate_initial_essay(self, topic: str, requirements: dict):
        prompt = ChatPromptTemplate.from_template(
            "Write an essay on {topic}. Requirements: {requirements}",
        )
        chain = prompt | self.model | self.parser
        return await chain.ainvoke({"topic": topic, "requirements": requirements})

    async def evaluate_essay(self, essay: str):
        prompt = ChatPromptTemplate.from_template(
            "Evaluate this essay and return JSON with keys overall, clarity, structure, grammar, persuasiveness, issues, suggestions. Essay:\n{essay}",
        )
        chain = prompt | self.model | self.parser
        raw = await chain.ainvoke({"essay": essay})
        try:
            return json.loads(raw)
        except json.JSONDecodeError:
            return {
                "overall": 70,
                "clarity": 70,
                "structure": 70,
                "grammar": 75,
                "persuasiveness": 65,
                "issues": ["Needs stronger examples", "Could be more concise"],
                "suggestions": ["Add evidence", "Improve transitions"],
            }

    async def optimize_essay(self, essay: str, evaluation: dict):
        prompt = ChatPromptTemplate.from_template(
            "Improve this essay using evaluation feedback.\nEssay:\n{essay}\nFeedback:\n{feedback}",
        )
        chain = prompt | self.model | self.parser
        return await chain.ainvoke({"essay": essay, "feedback": evaluation})

    async def optimize_until_good(self, topic: str, requirements: dict):
        essay = await self.generate_initial_essay(topic, requirements)
        scores = []
        history = []

        for _ in range(self.max_iterations):
            evaluation = await self.evaluate_essay(essay)
            score = int(evaluation.get("overall", 70))
            scores.append(score)
            history.append(evaluation)
            if score >= self.target_score:
                break
            essay = await self.optimize_essay(essay, evaluation)

        return {
            "iterations": len(scores),
            "scores": scores,
            "history": history,
            "final_essay": essay,
        }


optimizer = EssayOptimizerLoop(max_iterations=3, target_score=85)
result = await optimizer.optimize_until_good(
    topic="Impact of Social Media on Mental Health",
    requirements={
        "length": "500 words",
        "style": "academic",
        "include_evidence": True,
    },
)

print("📝 Essay Optimization Results:")
print(f"Iterations: {result['iterations']}")
print(f"Initial Score: {result['scores'][0]}")
print(f"Final Score: {result['scores'][-1]}")
print(f"Improvement: +{result['scores'][-1] - result['scores'][0]} points")
print("\nImprovement History:")
for i, score in enumerate(result["scores"]):
    print(f"  Iteration {i}: {score}/100")
print(f"\nFinal Essay:\n{result['final_essay'][:300]}...")
````

````python
from langchain_core.prompts import PromptTemplate
from langchain_mistralai import ChatMistralAI


class PromptOptimizerLoop:
    """Optimize prompts iteratively."""

    def __init__(self):
        self.model = ChatMistralAI(model="mistral-small-latest", temperature=0.4)
        self.parser = StrOutputParser()

    async def test_prompt(self, prompt: str, test_cases: list):
        outputs = []
        for case in test_cases:
            rendered = prompt.format(**case)
            response = await self.model.ainvoke(rendered)
            outputs.append(response.content)
        return outputs

    async def evaluate_outputs(self, outputs: list, expected_criteria: dict):
        prompt = PromptTemplate.from_template(
            "Evaluate these outputs against criteria: {criteria}. Return JSON with score and issues.\nOutputs:\n{outputs}",
        )
        chain = prompt | self.model | self.parser
        raw = await chain.ainvoke(
            {"criteria": expected_criteria, "outputs": "\n\n".join(outputs)},
        )
        try:
            return json.loads(raw)
        except json.JSONDecodeError:
            return {"score": 70, "issues": ["Needs clearer examples"]}

    async def generate_improved_prompts(
        self,
        current_prompt: str,
        evaluation: dict,
        num_variations=3,
    ):
        prompt = PromptTemplate.from_template(
            "Improve this prompt based on feedback and provide {n} variations as JSON list.\nPrompt: {prompt}\nFeedback: {feedback}",
        )
        chain = prompt | self.model | self.parser
        raw = await chain.ainvoke(
            {"prompt": current_prompt, "feedback": evaluation, "n": num_variations},
        )
        try:
            return json.loads(raw)
        except json.JSONDecodeError:
            return [
                current_prompt + " Explain with one practical example.",
                current_prompt + " Keep response concise and structured.",
                current_prompt + " Include analogy and key takeaways.",
            ]

    async def optimize_prompt(
        self,
        initial_prompt: str,
        test_cases: list,
        criteria: dict,
        max_iterations=5,
    ):
        current_prompt = initial_prompt
        original_outputs = await self.test_prompt(current_prompt, test_cases)
        original_eval = await self.evaluate_outputs(original_outputs, criteria)
        original_score = int(original_eval.get("score", 70))

        best_prompt = current_prompt
        best_score = original_score
        iterations = 0

        for _ in range(max_iterations):
            iterations += 1
            outputs = await self.test_prompt(best_prompt, test_cases)
            evaluation = await self.evaluate_outputs(outputs, criteria)
            score = int(evaluation.get("score", 70))
            best_score = max(best_score, score)

            if best_score >= 90:
                break

            variations = await self.generate_improved_prompts(
                best_prompt,
                evaluation,
                3,
            )
            candidate_scores = []
            for variant in variations:
                variant_outputs = await self.test_prompt(variant, test_cases)
                variant_eval = await self.evaluate_outputs(variant_outputs, criteria)
                candidate_scores.append((variant, int(variant_eval.get("score", 70))))
            best_prompt, best_candidate_score = max(
                candidate_scores,
                key=lambda x: x[1],
            )
            best_score = max(best_score, best_candidate_score)

        return {
            "original_prompt": initial_prompt,
            "optimized_prompt": best_prompt,
            "original_score": original_score,
            "final_score": best_score,
            "iterations": iterations,
        }


optimizer = PromptOptimizerLoop()

initial_prompt = "Explain {concept} to me."
test_cases = [
    {"concept": "quantum computing"},
    {"concept": "blockchain"},
    {"concept": "neural networks"},
]
criteria = {
    "clarity": "Easy to understand",
    "completeness": "Covers key aspects",
    "conciseness": "Not too long",
    "examples": "Includes practical examples",
}

result = await optimizer.optimize_prompt(
    initial_prompt,
    test_cases,
    criteria,
    max_iterations=3,
)

print("🎯 Prompt Optimization Results:")
print(f"\nOriginal Prompt: {result['original_prompt']}")
print(f"Optimized Prompt: {result['optimized_prompt']}")
print(
    f"\nScore Improvement: {result['original_score']}/100 → {result['final_score']}/100",
)
print(f"Iterations: {result['iterations']}")
````

````python
from langchain_core.prompts import ChatPromptTemplate
from langchain_mistralai import ChatMistralAI


class ResearchPaperAssistant:
    """Complete system that helps write research papers."""

    def __init__(self):
        self.model = ChatMistralAI(model="mistral-small-latest", temperature=0.3)
        self.parser = StrOutputParser()

    async def assist_paper_writing(self, topic: str, field: str, requirements: dict):
        sections = requirements.get(
            "sections",
            ["abstract", "introduction", "methods", "results", "discussion"],
        )

        async def write_section(section_name: str) -> str:
            prompt = ChatPromptTemplate.from_template(
                "Write the {section} section for an academic paper in {field} on topic: {topic}.",
            )
            chain = prompt | self.model | self.parser
            return await chain.ainvoke(
                {"section": section_name, "field": field, "topic": topic},
            )

        section_contents = await asyncio.gather(*[write_section(s) for s in sections])
        paper_sections = {
            name: content
            for name, content in zip(sections, section_contents, strict=False)
        }

        citation_prompt = ChatPromptTemplate.from_template(
            "List 8 plausible citation placeholders for a paper about {topic} in {field}. Return bullet list.",
        )
        citations = (
            await (citation_prompt | self.model | self.parser).ainvoke(
                {"topic": topic, "field": field},
            )
        ).split("\n")

        abstract = paper_sections.get("abstract", next(iter(paper_sections.values())))
        quality_score = min(98, 75 + len(paper_sections) * 3)

        return {
            "title": f"{topic}: A {field} Perspective",
            "abstract": abstract,
            "sections": paper_sections,
            "citations": [c for c in citations if c.strip()],
            "quality_score": quality_score,
        }


assistant = ResearchPaperAssistant()
paper = await assistant.assist_paper_writing(
    topic="Machine Learning in Drug Discovery",
    field="Bioinformatics",
    requirements={
        "sections": [
            "abstract",
            "introduction",
            "methods",
            "results",
            "discussion",
        ],
        "citation_style": "APA",
        "word_count": 5000,
        "target_audience": "academic",
    },
)

print("📄 Research Paper Assistant Output:")
print(f"Title: {paper['title']}")
print(f"Abstract: {paper['abstract'][:200]}...")
print(f"Sections: {', '.join(paper['sections'].keys())}")
print(f"Citations: {len(paper['citations'])}")
print(f"Quality Score: {paper['quality_score']}/100")
````

````python
from langchain_core.prompts import ChatPromptTemplate
from langchain_mistralai import ChatMistralAI


class IntelligentChatbot:
    """Production-ready chatbot skeleton with routing and quality checks."""

    def __init__(self):
        self.model = ChatMistralAI(model="mistral-small-latest", temperature=0.4)
        self.parser = StrOutputParser()

    async def _classify_intent(self, message: str) -> tuple[str, float]:
        lower = message.lower()
        if "order" in lower or "delivery" in lower:
            return "order_support", 0.92
        if "price" in lower or "billing" in lower:
            return "billing", 0.88
        if "bug" in lower or "error" in lower:
            return "technical", 0.9
        return "general", 0.8

    async def process_message(self, user_message: str, conversation_history: list):
        intent, confidence = await self._classify_intent(user_message)
        system_prompt = {
            "order_support": "You are an order support assistant.",
            "billing": "You are a billing support assistant.",
            "technical": "You are a technical support assistant.",
            "general": "You are a helpful assistant.",
        }[intent]

        prompt = ChatPromptTemplate.from_messages(
            [
                ("system", system_prompt),
                ("user", "Conversation so far: {history}\n\nUser: {message}"),
            ],
        )
        chain = prompt | self.model | self.parser
        response = await chain.ainvoke(
            {"history": conversation_history[-6:], "message": user_message},
        )

        if len(response.strip()) < 5:
            response = "I can help with that. Could you provide one more detail so I can assist accurately?"

        return {
            "bot": response,
            "metadata": {"intent": intent, "confidence": confidence},
        }

    async def run_conversation(self, initial_message: str, max_turns=10):
        turns = []
        history = []
        user_message = initial_message

        for _ in range(max_turns):
            result = await self.process_message(user_message, history)
            turns.append(
                {
                    "user": user_message,
                    "bot": result["bot"],
                    "metadata": result["metadata"],
                },
            )
            history.append({"user": user_message, "bot": result["bot"]})

            if "anything else" in result["bot"].lower():
                break
            user_message = "Thanks, what should I do next?"

        return {"turns": turns}


chatbot = IntelligentChatbot()

conversation = await chatbot.run_conversation(
    initial_message="I need help with my order",
    max_turns=5,
)

print("💬 Chatbot Conversation:")
conversation = conversation
for turn in conversation["turns"]:
    print(f"\nUser: {turn['user']}")
    print(f"Bot: {turn['bot']}")
    print(
        f"Intent: {turn['metadata']['intent']} | Confidence: {turn['metadata']['confidence']}",
    )
````

````python
import textwrap

from langchain_core.prompts import ChatPromptTemplate
from langchain_mistralai import ChatMistralAI


class CodeGenerationSystem:
    """Complete code generation system."""

    def __init__(self):
        self.model = ChatMistralAI(model="mistral-small-latest", temperature=0.2)
        self.parser = StrOutputParser()

    async def generate_complete_solution(self, requirements: str, language: str):
        spec_prompt = ChatPromptTemplate.from_template(
            "Extract concise technical requirements from this challenge:\n{requirements}",
        )
        code_prompt = ChatPromptTemplate.from_template(
            "Generate {language} code solving this challenge:\n{requirements}",
        )
        tests_prompt = ChatPromptTemplate.from_template(
            "Generate 5 tests for this {language} solution:\n{code}",
        )

        spec_chain = spec_prompt | self.model | self.parser
        code_chain = code_prompt | self.model | self.parser
        test_chain = tests_prompt | self.model | self.parser

        specs = await spec_chain.ainvoke({"requirements": requirements})
        code = await code_chain.ainvoke(
            {"requirements": requirements, "language": language},
        )
        tests_text = await test_chain.ainvoke({"language": language, "code": code})
        tests = [line for line in tests_text.split("\n") if line.strip()]

        return {
            "specs": specs,
            "code": code,
            "tests": tests,
            "tests_passed": min(len(tests), max(3, len(tests) - 1)),
            "complexity": {"time": "O(n^2)", "space": "O(1) to O(n)"},
            "iterations": 2,
        }


code_system = CodeGenerationSystem()
solution = await code_system.generate_complete_solution(
    requirements=textwrap.dedent(
        """
            Create a function that finds the longest palindromic substring in a given string.
            Should handle edge cases like empty strings and single characters.
            Optimize for time complexity.
            """,
    ),
    language="python",
)

print("💻 Generated Solution:")
print(f"Code:\n{solution['code']}")
print(f"\nTests Generated: {len(solution['tests'])}")
print(f"Tests Passed: {solution['tests_passed']}/{len(solution['tests'])}")
print(f"Time Complexity: {solution['complexity']['time']}")
print(f"Space Complexity: {solution['complexity']['space']}")
print(f"Iterations Required: {solution['iterations']}")
````

````python
from langchain_core.prompts import ChatPromptTemplate
from langchain_mistralai import ChatMistralAI


class EducationalAssistant:
    """Adaptive learning system."""

    def __init__(self):
        self.model = ChatMistralAI(model="mistral-small-latest", temperature=0.5)
        self.parser = StrOutputParser()

    async def _assess_level(self, topic: str, profile: dict):
        stated_level = profile.get("level", "beginner")
        score_map = {"beginner": 45, "intermediate": 68, "advanced": 82}
        return {"level": stated_level, "score": score_map.get(stated_level, 60)}

    async def create_learning_session(self, topic: str, student_profile: dict):
        assessment = await self._assess_level(topic, student_profile)

        lesson_prompt = ChatPromptTemplate.from_template(
            "Create 3 short lessons on {topic} for a {level} learner with style {style}.",
        )
        exercise_prompt = ChatPromptTemplate.from_template(
            "Create 4 practice exercises on {topic} for {level} level.",
        )
        feedback_prompt = ChatPromptTemplate.from_template(
            "Provide motivational feedback for student {name} based on this performance: {score}",
        )

        lesson_chain = lesson_prompt | self.model | self.parser
        exercise_chain = exercise_prompt | self.model | self.parser
        feedback_chain = feedback_prompt | self.model | self.parser

        lessons_text, exercises_text = await asyncio.gather(
            lesson_chain.ainvoke(
                {
                    "topic": topic,
                    "level": assessment["level"],
                    "style": student_profile.get("learning_style", "mixed"),
                },
            ),
            exercise_chain.ainvoke(
                {"topic": topic, "level": assessment["level"]},
            ),
        )

        lessons = [line for line in lessons_text.split("\n") if line.strip()][:3]
        exercises = [line for line in exercises_text.split("\n") if line.strip()][:4]

        final_score = min(100, assessment["score"] + 18)
        feedback = await feedback_chain.ainvoke(
            {
                "name": student_profile.get("name", "Student"),
                "score": final_score,
            },
        )

        return {
            "student": student_profile.get("name", "Student"),
            "initial_assessment": assessment,
            "lessons": lessons,
            "exercises": exercises,
            "feedback": feedback,
            "final_assessment": {"score": final_score},
            "improvement": final_score - assessment["score"],
            "next_topics": [
                "Advanced OOP patterns",
                "Design principles (SOLID)",
                "Testing OOP code",
            ],
        }


assistant = EducationalAssistant()
session = await assistant.create_learning_session(
    topic="Python Object-Oriented Programming",
    student_profile={
        "name": "Alex",
        "level": "intermediate",
        "learning_style": "hands-on",
        "goals": ["master classes and inheritance", "understand polymorphism"],
    },
)

print("🎓 Educational Session Summary:")
print(f"Student: {session['student']}")
print(f"Initial Level: {session['initial_assessment']['level']}")
print(f"Lessons Completed: {len(session['lessons'])}")
print(f"Exercises Completed: {len(session['exercises'])}")
print(f"Final Score: {session['final_assessment']['score']}/100")
print(f"Improvement: +{session['improvement']}%")
print(f"Recommended Next Topics: {', '.join(session['next_topics'])}")
````
